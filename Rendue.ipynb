# Etape a suivre pour resoudre le projet

Pour faire en sorte de pouvoir simplement prendre une photo de ce que on as écrit pour que ensuite il nous renvoie un txt ou word de ce que on a écrit, il faut commencer par prendre une photo.

Avec cette photo comme on compte utiliser la database EMNIST et que cette base de donnée suis certaine reglementation alors on devra les suivre : Nuances de noir et blanc, et image d'une lettre de dimension 28 * 28 px

Pour atteindre ses contraintes, il faut : transformer le png en noir et blanc , segmenter les mots des espaces pour dire ca c'est un mot, ca c'est un espace...
puis pour chaque mots il faudra les segmenter en lettre, c'est a dire que il faut dire que un mots est une association de lettre, en revanche les lettres prendront differentent taille apres segmentation 40 * 60 ou peut etre 100 * 300 donc il faudra ensuite les redimmensionner en 28 * 28 afin de les passer une par une dans le modele basé sur EMNIST afin de ensuite remonter la chaine pour former les mots puis les espaces puis l'entiereter du documents.

Voici la base du projet
