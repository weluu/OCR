```python
import cv2
import numpy as np

def segmenter_mots(image_path):
    # 1. Charger l'image en nuances de gris
    image = cv2.imread(image_path)
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # 2. Binarisation (inversion pour avoir le texte en blanc sur fond noir)
    # On utilise OTSU pour calculer automatiquement le meilleur seuil de contraste
    ret, thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

    # 3. Dilatation (L'étape clé)
    
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (50, 50)) # 50 : largeur , 50: hauteur 
    dilated = cv2.dilate(thresh, kernel, iterations=1)

    # 4. Détection des contours
    # Chaque contour trouvé correspond normalement à un mot
    contours, _ = cv2.findContours(dilated, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    print(f"{len(contours)} mots potentiels détectés.")

    for i, cnt in enumerate(contours):
        x, y, w, h = cv2.boundingRect(cnt)
        
        # Filtrer les bruits 
        
        if w > 20 and h > 10:
            # On découpe le mot original
            mot_decoupe = image[y:y+h, x:x+w]
            
            # Sauvegarder ou afficher le mot
            cv2.imwrite(f'mot_{i}.png', mot_decoupe)
            
            # Dessiner un rectangle sur l'image d'origine pour visualiser -> etape pour les tests 
            cv2.rectangle(image, (x, y), (x + w, y + h), (255, 255, 0), 3)

    # Affichage du résultat final
    cv2.imshow('Mots detectes', image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

# Utilisation
segmenter_mots('sentence3.png')
```
